## Character occurence in a string/lst

```python
#method 1 (BestOne)
def character_occurence(s):
    counts = dict()
    for i in s : 
        if i not in counts:
            counts[i] = 0
        counts[i] += 1 
    return(counts)



result_dict = character_occurence('aabbababacacccagttttkkkssd')
print(result_dict)

#finding least occuring element
print(min(result_dict, key=result_dict.get))
```

```python
#method 2 (inbuilt Counter function)
from collections import Counter
def character_occurence(s):
    counts = Counter(s)
    return counts


result_dict = character_occurence('aabbababacacccagttttkkkssd')
print(result_dict)
#finding least occuring element
print(min(result_dict, key=result_dict.get))
```

<h4>finding count</h4>

```python
#method 1 
def counter(s, ele):
    return s.count(ele)

counter('aabbababacacccagttttkkkssd','a')
```

```python
#method 2
def counter(s, ele):
    c = 0
    for i in s :
        if i == ele :
            c += 1 
    return c

counter('aabbababacacccagttttkkkssd','a')
```

```python
#method 3
def counter(s, ele):
    counts = dict()
    for i in s : 
        if i not in counts:
            counts[i] = 0
        counts[i] += 1 
    return(counts[ele])

counter('aabbababacacccagttttkkkssd','a')
```

```python
#method 4
from collections import Counter
def character_occurence(s, w):
    counts = Counter(s)
    return counts[w]


counter('aabbababacacccagttttkkkssd','a')
```
